```python
dataset = {
    'cars' : ['BMW', 'Porsche', 'Nissan', 'Opel', 'Ferrari', 'Hyunday'],
    'counts' : [3, 7, 2, 2, 5, 9],
    'years' : [2024, 2026, 2023, 2020, 2025, 2025],
    'origins' : ['Ger', 'Ger', 'Jap', 'Ger', 'Ita', 'Kor']
}


dataFrame = pd.DataFrame(dataset)
print('dataFrame: ', dataFrame)
```
