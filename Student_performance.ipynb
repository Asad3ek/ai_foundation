```python
df['total productivity'] = df[['Study_Hours_Per_Week', 'Attendance_Percentage', 'Previous_Score']].sum(axis=1)
df['avarage productivity'] = df['total productivity'] /3
conditions = [
    df['avarage productivity'] >= 65,
    (df['avarage productivity'] >= 55) & (df['avarage productivity'] < 65)
]

choices = ['Good', 'Norm']
df['productive rate'] = np.select(conditions, choices, default='Bad')

print(df[['avarage productivity', 'productive rate']])
```

```python
Internet_Access =df.groupby('Internet_Access')[['Study_Hours_Per_Week', 'Previous_Score']].mean()

print('student previous score by internet access: ')
print(Internet_Access)
```

```python
#avarage score of students by age
Student_Age =df.groupby('Age')[['Study_Hours_Per_Week', 'Previous_Score']].mean()

print('student previous score by Age: ')
print(Student_Age)

print('=======================================================================')

#avarage score of students by gender
Student_Gender =df.groupby('Gender')[['Study_Hours_Per_Week', 'Previous_Score']].mean()

print('student previous score by Gender: ')
print(Student_Gender)
```

```python
#check passed studends vs failed
df['total score'] = df[['Study_Hours_Per_Week','Previous_Score']].sum(axis=1)
df['avarage_score'] = df['total productivity'] /2
df['who_passed'] = np.where(df['avarage_score'] >= 65, 'Yes', 'No')
pass_rate = df['who_passed'].value_counts(normalize=True) * 100


print( pass_rate)
```

```python
#show all passed and failed students.

df['total_score'] = df[['Study_Hours_Per_Week','Previous_Score']].sum(axis=1)
df['avarage_score'] = df['total_score'] /2
condition =[
    df['avarage_score'] >= 50,
    df['avarage_score'] < 50
     ]
boolean = ["PASSED", "FAILED"]

df['exam_score'] = np.select(condition, boolean, default='FAILED')
print(df[['avarage_score', 'exam_score']])
```

```python
df['Speaking Score'] = np.random.choice([5.0, 5.5, 6.0, 6.5, 7.0, 7.5, 8.0, 8.5, 9.0], size=10)

print(df)
```

```python
df[['Age', 'Listening Score', 'Reading Score', 'Writing Score', 'Speaking Score']]
```

```python
total_score = df[['Listening Score', 'Reading Score', 'Writing Score', 'Speaking Score']].sum(axis=1)
average_score = df['total_score'] / 4
averaged_score = df['average_score'].round(1)

print('averaged_score:\n', averaged_score)
print('---------------------------------')

#university require min score is 6.5
passed_students = df[df['averaged_score'] >= 6.5]
failed_students = df[df['averaged_score'] < 6.5]

#percentage of passed students
percentage = (df['average_score'] >= 6.5).mean() * 100

print('Students who gained enough score : ')
print(passed_students[['Student_ID', 'averaged_score']])
print('---------------------------------')
print('Students who could not gain enough score : ')
print(failed_students[['Student_ID', 'averaged_score']])
```

```python
#convert scores to numpy array
scores = df[['Listening Score', 'Reading Score', 'Writing Score', 'Speaking Score']].values

print('numpy mean: ', np.mean(scores, axis=0))
print('numpy std daviation: ', np.std(scores, axis=0))

print('------------------------------------')
#z-score(normalize)
z_score = (scores - np.mean(scores, axis=0)) / np.std(scores, axis=0)
print('z-score(normalized numbers)')
print(z_score)
```

```python
#how many students scored above 7?

top_scorers = df[df['averaged_score'] >= 7 ]
print('Top Scored Students: ', len(top_scorers))
print('Top Scored Students info: ')
top_scorers[['Listening Score', 'Reading Score', 'Writing Score',
             'Speaking Score']]


#how many students failed?
failed = df[df['averaged_score'] <7]
print('Failed Students: ', len(failed))
print('Failed Students info: ')
failed[['Listening Score', 'Reading Score', 'Writing Score', 'Speaking Score']]
```
